# Part 1: Sources of bias in datasets

In class, we discussed different sources of bias that can affect the data we want to use for our Data Science applications. Here is a summary:

### 1. Historical bias
**Historical bias:** bias that exists in society and is reflected in the data. It is the most insidious because it arises even if we are able to perfectly sample from the existing population. Most often, it affects groups that are historically disadvantaged.

E.g. In 2018, 5% of Fortune 500 CEOs were women. Historically, women have less frequently made it to a CEO position. A classifier trained to predict the best choice for a new CEO may learn this pattern and determine that being a woman makes one less qualified to be a CEO.

### 2. Representation bias
**Representation bias:** the sample underrepresents part(s) of the population and fails to generalize well. This may happen for different reasons:

1. The sampling methods only reached a portion of the population. E.g. Data collected via smartphone apps can under-represent lower incomes or older groups, who may be less likely to own smartphones.

2. The population of interest has changed or is distinct from the sample used during model training. E.g. Data that is representative of Vancouver may not be representative if used to analyze the population in Toronto. Similarly, data representative of Vancouver 20 years ago may not reflect today's population. 

### 3. Measurement bias
**Measurement bias:** it occurs when choosing features that fail to correctly represent the problem, or when there are issues with the data collection. Fore example:

1. The measurement processes varies across groups. E.g. one group of workers is monitored more closely and thus more errors are observed in that group.

2. The quality of data varies across groups. E.g. women often self-report less pain than men and are therefore less likely to receive certain diagnoses

3. The defined classification task or one of the features used is an oversimplification. E.g. We are designing a model to predict whether a student will be successful in college. We choose to predict the final GPA as metric of success. This, however, ignores other indicators of success.

**WEB SCRAPING SUMMARY**

<u>Legal when:</u> data is publicly accessible, non-personal and does not violate laws, contracts, or technical access restrictions; doesn’t require the web scraper to log in and explicitly accept the website’s terms of conditions

<u>Illegal when:</u> it involves personal data, copyrighted content, or restricted access, or when it breaches applicable laws or terms of service; need to log in to access data and terms and conditions forbid web scraping

**Personal data**: Name, email, phone number, address, username, IP address, DOB, employment, bank, medical, biometric

**Lawful reasons to collect data**
1. Consent – the data subject consented to us having their data.
2. Contract – personal data is required to perform a contract with the data subject.
3. Compliance – necessary for compliance with a legal obligation.
4. Vital Interest, Public Interest, or Official Authority – typically only applicable for state-run bodies where access to personal data is in the public’s interest.
5. Legitimate Interest

**Copyrighted data**: Articles, Videos, Pictures, Stories, Music, Databases

**Fair use** permits limited use of copyrighted material without needing permission from copyright owner. Facts like product names, prices, features are not copyrightable

**When scraping databases (for ethical purposes):** only scrape some of the available data, do not replicate the organizational structure of the original database

<u>Applicable Laws</u>
1. GDPR (General Data Protection Regulation; EU law)
If scraping any personal data from European users, need:
- Explicit consent or
- Legitimate interest that outweighs the privacy rights of the individuals involved

If you’re scraping personal information about Californians, need to:

- Prohibits “unauthorized access” to computer systems

## Observational and experimental studies

- **Observational study:** study where there is no deliberate human intervention regarding the variable under investigation. Observational studies are ones where researchers observe the effect of a treatment/intervention without trying to change who is or isn’t exposed to it. In an observational study, the subjects are assigned or assign themselves to the exposure group they belong to.
- **Experimental study:** study that involves planned intervention on the exposure to a condition. In an experiment, subjects are assigned to a condition by the researcher and thus one can establish a cause-and-effect relationship when we see a difference in the outcome between the experimental groups. Randomizing study subjects balances any differences between treatment groups with respect to all variables except the condition of exposure.

## A/B testing

A/B testing can be considered the most basic kind of randomized controlled experiment. 

Complete the following reading, then answer the comprehension questions below: https://hbr.org/2017/06/a-refresher-on-ab-testing

**A/B TESTING SUMMARY**

**A/B testing** is a quantitative research method that tests two or more design variations with a live audience to determine which variation performs best according to a predetermined set of business-success metrics.
- compare original design A (control version) and one variation B (variant) that differs from control only in one design element
- in most cases, if variant statistically significantly outperforms the original design, it should be used as the new design of the product

<u> A/B Testing Setup </u>
1. Hypothesis - based on user research and business insights
2. Define changes to make - only address just one design element
3. Choose outcome metrics - should define **primary metrics**, which will tell you if the design variation results in the hoped-for change in behavior, and **guardrail metrics**, to determine if the change in user behavior truly has a positive impact on the business
4. Determine time frame of test - depends on required *sample size* for the test; heuristic is to run for 1-2 weeks to account for potential fluctuations in user behavior

<u> Determining Sample Size </u>
- Baseline outcome-metric value: The outcome metric (e.g., conversion rate, click rate) for  your design
- Minimum detectable effect: The minimum relative change in the outcome metric that  you want to be able to detect
- Statistical-significance threshold (usually 95%): The degree of certainty you want to have that your test result is reliable

<u> Limitations </u>
- Not good for low-traffic pages (sample size too small)
- Don't test multiple changes at a time --> lack an understanding of the impact of each individual change; perform multivariate testing instead (requires even larger sample size)
- Doesn't provide why behind change in behaviour

**Question 4**

Explain the role of blocking in A/B testing.

Blocking is the process of grouping experimental units into homogeneous subgroups (blocks) based on a characteristic you expect to influence the outcome, and then randomizing treatment assignment within each block rather than across the whole sample.

For example: a website tests a new checkout button. Mobile users convert at 2% and desktop users at 6%. Without blocking, a random excess of desktop users in one arm could make that variant look better regardless of the button. Blocking on device ensures both arms have the same mobile/desktop mix, so differences in conversion can be attributed to the button.

**Question 5**

The authors warn about observing too many metrics when running an A/B test. Why is that the case? What could happen if I ignore this warning?

Observing too many metrics in an A/B test increases false positives through random chance. At a significance level of α = 0.05, testing 20 metrics gives roughly a 64% (1 − 0.95²⁰) chance that at least one appears significant even if the change has no effect. It also creates unclear trade-offs when different metrics move in opposite directions, which invites cherry-picking and post-hoc storytelling (constructing an explanation for the results after they've been seen). 

To avoid this, the experimenter should pre-specify a primary metric, treat the others as secondary or guardrail metrics, and correct for multiple comparisons.

### Ethical A/B testing
Ethical A/B testing still requires all the ethical considerations of any experimental study, such as informed consent or possibility to opt out. A notorious case of a company failing to meet ethics requirement in A/B testing is the infamous Facebook "social contagion experiment", in which almost 700,000 users were showed, for a week, only positive or only negative content, to see how this variation impacted their online behaviour. The selected users were not informed and could not opt out. Furthermore, their emotional state was affected. Facebook defended itself by saying that Facebook's Data Use Policy warns users that Facebook “may use the information we receive about you […] for internal operations, including troubleshooting, data analysis, testing, research and service improvement”. This defense was largely rejected by the scientific community, which still considered the study as unethical. You can read more about this incident in this [article](https://www.theguardian.com/technology/2014/jun/30/facebook-emotion-study-breached-ethical-guidelines-researchers-say).

# Part 2: Privacy



When collecting data for a study, privacy is almost always a primary concern. Our data set may include information that makes it possible to identify an individual, including:

- **Direct identifiers**, which are the ones that can be used to uniquely identify an individual or a household in a dataset, such as a record ID number, patient number, social insurance number, full address, etc. Usually, name is also considered a direct identifier (although several people can have the same name). Other features such as age, date of birth, or postal code are not sufficient on their own to uniquely identify an individual and would not be considered direct identifiers.
- **Indirect (or quasi) identifiers**, which are the columns that do not themselves identify any individual or household, but can do so when combined with other indirect-identifiers. For example, postal code and date of birth are often indirect identifiers, because it is very likely that within a zip code only one individual has this particular birth date. The more indirect identifiers that you have, the more likely it is that individuals become identifiable because there are more possible unique combinations of identifying features.

A big issue with suppression of direct identifier is that it is not reversible. If at some point we need to identify an individual in our dataset, we would be out of luck. If you have reasons to believe that re-identification may be required, pseudonymization would be a better option to handle direct identifiers. Pseudonymization replaces one or more direct identifiers with a unique but less meaningful value. Usually when we pseudonymize an identifier, there is a possibility of re-identification if required (but it would not be available to the general public).

**Question 11:** pseudomyze the ID column of the NIJ training set and save the result in a new dataframe called `pseudo_df`. In a different code cell, show that it is possible to re-identify the samples by converting them back to the original ID number.

There are different ways to achieve this you may want to explore:

- Use an extisting library, such as [`cryptography`](https://cryptography.io/en/latest/).

````python
# Your answer here (you may add more cells as needed)

SECRET_XOR = 0x5A5A5A
SECRET_MULT = 7919
SECRET_ADD = 123456789
MOD = 2**32

def pseudonymize_id(original_id):
    return ((int(original_id) ^ SECRET_XOR) * SECRET_MULT + SECRET_ADD) % MOD

def reidentify_id(pseudo_id):
    inv_mult = pow(SECRET_MULT, -1, MOD)
    return (((int(pseudo_id) - SECRET_ADD) * inv_mult) % MOD) ^ SECRET_XOR

# Pseudonymize
pseudo_df = df.copy()
pseudo_df["ID"] = pseudo_df["ID"].apply(pseudonymize_id)
pseudo_df[["ID"]].head()
````

````python
# Re-identifying IDs
recovered_ids = pseudo_df["ID"].apply(reidentify_id)

print((recovered_ids == df["ID"]).all())   # should print True
pd.DataFrame({"original": df["ID"].head(),
              "pseudonym": pseudo_df["ID"].head(),
              "recovered": recovered_ids.head()})
````

Remember that now, when checking for $k$-anonymity, you should be looking at the new column `Binned_PUMA`, not at `Residence_PUMA`.

````python
# Your answer here
df["Residence_PUMA"].describe()

base_qi = ["Gender", "Race", "Age_at_Release"]

for n_bins in range(25, 0, -1):
    binned = pd.qcut(df["Residence_PUMA"], q=n_bins,
                     labels=False, duplicates="drop")
    temp = df.copy()
    temp["Binned_PUMA"] = binned
    k = anonymity.k_anonymity(temp, base_qi + ["Binned_PUMA"])
    # print(n_bins, "bins -> k =", k)
    if k == 4:
        break

best_bins = n_bins

df["Binned_PUMA"] = pd.qcut(df["Residence_PUMA"], q=best_bins,
                                  labels=False, duplicates="drop")

k = anonymity.k_anonymity(df, ["Gender", "Race", "Age_at_Release", "Binned_PUMA"])
print(k)
````
